```python
game_ids = df['gameId'].astype(str).str.zfill(10).unique().tolist()
game_ids
```

```python
from nba_api.stats.endpoints import boxscoresummaryv2
import time

results = {}
for gid in game_ids:
    box = boxscoresummaryv2.BoxScoreSummaryV2(game_id=gid)
    line_score = box.line_score.get_data_frame()
    results[gid] = line_score[['TEAM_ABBREVIATION', 'TEAM_CITY_NAME', 'PTS']]
    print(gid)
    print(results[gid])
    print()
    time.sleep(2)
```

```python
results
```

```python
corrections = {}
for gid, line_score in results.items():
    rows = line_score.reset_index(drop=True)
    team0_pts = rows.loc[0, 'PTS']
    team1_pts = rows.loc[1, 'PTS']
    team0_abbr = rows.loc[0, 'TEAM_ABBREVIATION']
    team1_abbr = rows.loc[1, 'TEAM_ABBREVIATION']
    
    corrections[(gid, team0_abbr)] = {'teamScore': team0_pts, 'opponentScore': team1_pts}
    corrections[(gid, team1_abbr)] = {'teamScore': team1_pts, 'opponentScore': team0_pts}

corrections
```

```python
for (gid, team_abbr), vals in corrections.items():
    mask = (team_stats_ext['gameId'].astype(str).str.zfill(10) == gid)
    team_stats_ext.loc[mask, 'teamScore'] = vals['teamScore']
    team_stats_ext.loc[mask, 'opponentScore'] = vals['opponentScore']
    team_stats_ext.loc[mask, 'plusMinusPoints'] = vals['teamScore'] - vals['opponentScore']
```

```python
# Define a function to map a date to its NBA season "start year"
def get_season_start_year(date):
    if date.month >= 10:  # Oct, Nov, Dec -> season starts this year
        return date.year
    else:  # Jan-June (and any stray summer dates) -> season started previous year
        return date.year - 1
```

```python
team_stats_ext['gameDateTimeEst'].describe()
```

```python
team_stats_ext['gameDate'] = pd.to_datetime(team_stats_ext['gameDateTimeEst'], errors='coerce').dt.date
```

```python
team_stats_ext['gameDate'].describe()
```

```python
team_stats_ext['season'] = team_stats_ext['gameDate'].apply(get_season_start_year).astype('Int64')
```
